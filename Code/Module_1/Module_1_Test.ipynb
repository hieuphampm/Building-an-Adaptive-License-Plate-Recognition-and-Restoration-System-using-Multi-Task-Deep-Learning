```python
from ultralytics import YOLO
import os

model_path = "/content/drive/MyDrive/Building-an-Adaptive-License-Plate-Recognition-and-Restoration-System-using-Multi-Task-Deep-Learning/Training/Module_1/Model/module_1/weights/best.pt"
model = YOLO(model_path)
```

```python
image_folder = "/content/drive/MyDrive/Building-an-Adaptive-License-Plate-Recognition-and-Restoration-System-using-Multi-Task-Deep-Learning/Datasets/Datasets/module1/mock"
SAVE_DIR = "/content/drive/MyDrive/Building-an-Adaptive-License-Plate-Recognition-and-Restoration-System-using-Multi-Task-Deep-Learning/Training/Module_1/Test/Output_img_1"

results = model.predict(
    source=image_folder,
    save=True,
    save_txt=True,
    conf=0.25,
    imgsz=512,
    show=False,
    project=SAVE_DIR,
    name="",
    exist_ok=True
)

print("Detection done. Saved in:", os.path.abspath(SAVE_DIR))
```
