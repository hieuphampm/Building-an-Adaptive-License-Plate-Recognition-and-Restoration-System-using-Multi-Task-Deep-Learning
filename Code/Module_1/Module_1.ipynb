```python
DATA = "/content/drive/MyDrive/Building-an-Adaptive-License-Plate-Recognition-and-Restoration-System-using-Multi-Task-Deep-Learning/Datasets/Datasets/module1"
SAVE_DIR = "/content/drive/MyDrive/Building-an-Adaptive-License-Plate-Recognition-and-Restoration-System-using-Multi-Task-Deep-Learning/Training/Module_1/Model"
```

```python
import os

YAML_PATH = '/content/drive/MyDrive/Building-an-Adaptive-License-Plate-Recognition-and-Restoration-System-using-Multi-Task-Deep-Learning/Datasets/Datasets/module1/data.yaml'

if os.path.exists(YAML_PATH):
  print("data.yaml found")
  !cat "$YAML_PATH"
else:
  print("Not found data.yaml")
```

```python
from ultralytics import YOLO

model = YOLO('yolov8n.pt')
model.train(
    data = YAML_PATH,
    epochs = 150,
    imgsz=512,
    batch=16,
    name='module_1',
    project = SAVE_DIR,
    exist_ok = True,
    pretrained=True,
    patience=15,
)
```

```python
best_model_path = f"{SAVE_DIR}/module_1/weights/best.pt"
best_model = YOLO(best_model_path)

metrics = best_model.val()
print(f"mAP@50: {metrics.box.map50:.3f} | mAP@50-95: {metrics.box.map:.3f}")
```
